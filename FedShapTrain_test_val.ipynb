```python
# ============================================================
# Federated Learning (FedAvg) + FedShapley (approx) for:
# 1) Motor_Symptoms
# 2) Disease_Stage
# 3) Category
# Includes: overfit/underfit check + solutions + comparison table
# ============================================================

# ---------------------------
# Install / Imports
# ---------------------------
!pip -q install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cpu

import os
import copy
import random
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ---------------------------
# Load dataset
# (Use your preprocessed scaled file if you want)
# ---------------------------
# If you already mounted drive, keep your df read. Otherwise you can use this path:
path = "/content/drive/MyDrive/Dataset/Huntington_Disease_Dataset_Preprocessed_Scaled.csv.gz"  # provided uploaded file path
df = pd.read_csv(path)
print("Dataset loaded:", df.shape)

# ---------------------------
# Utilities
# ---------------------------
def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

class TabDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

class MLP(nn.Module):
    def __init__(self, input_dim, num_classes, hidden=128, dropout=0.25):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden, hidden//2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden//2, num_classes)
        )

    def forward(self, x):
        return self.net(x)

def get_metrics(y_true, y_pred):
    return {
        "Accuracy":  accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, average='weighted', zero_division=0),
        "Recall":    recall_score(y_true, y_pred, average='weighted', zero_division=0),
        "F1":        f1_score(y_true, y_pred, average='weighted', zero_division=0)
    }

@torch.no_grad()
def eval_model(model, X, y, batch_size=512):
    model.eval()
    ds = TabDataset(X, y)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False)
    preds = []
    for xb, _ in dl:
        xb = xb.to(device)
        logits = model(xb)
        pred = torch.argmax(logits, dim=1).cpu().numpy()
        preds.append(pred)
    y_pred = np.concatenate(preds)
    return y_pred, get_metrics(y, y_pred)

def train_one_client(model, loader, epochs=1, lr=1e-3, weight_decay=1e-4):
    model.train()
    opt = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    loss_fn = nn.CrossEntropyLoss()
    for _ in range(epochs):
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad()
            logits = model(xb)
            loss = loss_fn(logits, yb)
            loss.backward()
            opt.step()

def fedavg_aggregate(global_model, client_models, client_sizes):
    # Weighted average of client parameters
    global_dict = global_model.state_dict()
    total = sum(client_sizes)

    new_state = {}
    for k in global_dict.keys():
        new_state[k] = torch.zeros_like(global_dict[k])

    for cm, sz in zip(client_models, client_sizes):
        cdict = cm.state_dict()
        w = sz / total
        for k in new_state.keys():
            new_state[k] += cdict[k] * w

    global_model.load_state_dict(new_state)

def split_into_clients(X_train, y_train, n_clients=5):
    # simple shuffle split (stratified-ish by mixing)
    idx = np.arange(len(y_train))
    np.random.shuffle(idx)
    splits = np.array_split(idx, n_clients)
    clients = []
    for s in splits:
        clients.append((X_train[s], y_train[s]))
    return clients

def overfit_underfit_diagnosis(train_acc_hist, val_acc_hist):
    # Very simple diagnostic:
    # - Overfit: train improves much more than val and val plateaus/drops
    # - Underfit: both low and close together, not improving much
    if len(train_acc_hist) < 3:
        return "Not enough rounds to diagnose reliably."

    train_last = train_acc_hist[-1]
    val_last   = val_acc_hist[-1]
    gap = train_last - val_last

    if train_last > 0.90 and gap > 0.08:
        return ("Likely OVERFITTING: train accuracy is high but validation lags.\n"
                "Fixes: increase dropout/weight_decay, reduce model size, early stopping, "
                "more data (or stronger regularization), fewer rounds/local epochs.")
    if train_last < 0.65 and val_last < 0.65 and abs(gap) < 0.03:
        return ("Likely UNDERFITTING: both train and validation are low and similar.\n"
                "Fixes: increase model capacity, train longer (more rounds/local epochs), "
                "tune LR (try 1e-3 to 3e-3), reduce regularization, add better features.")
    if gap <= 0.08:
        return ("Looks OK: train and validation are reasonably aligned.\n"
                "If you want higher accuracy: tune hyperparams (rounds, LR, hidden size), "
                "and consider class balancing or better feature engineering.")
    return ("Mixed signals: consider tuning rounds/local epochs/LR and use early stopping.")

# ------------------------------------------------------------
# FedShapley (approx via Monte-Carlo permutations)
# Utility = validation accuracy after a small FL training budget
# ------------------------------------------------------------
def fedshapley_approx(
    X_train, y_train, X_val, y_val,
    input_dim, num_classes,
    n_clients=5,
    mc_permutations=10,
    budget_rounds=3,
    local_epochs=1,
    lr=1e-3,
    weight_decay=1e-4,
    batch_size=128,
    hidden=128,
    dropout=0.25
):
    """
    Approx Shapley for clients:
    For each random permutation, we add clients one-by-one and measure marginal gain
    in validation accuracy after a small FL training budget on the subset.
    Kept small to run in Colab.
    """
    set_seed(SEED)

    clients = split_into_clients(X_train, y_train, n_clients=n_clients)
    base_init = MLP(input_dim, num_classes, hidden=hidden, dropout=dropout).to(device)
    base_state = copy.deepcopy(base_init.state_dict())

    def train_subset(client_indices):
        # Train FedAvg on the subset for a small budget
        global_model = MLP(input_dim, num_classes, hidden=hidden, dropout=dropout).to(device)
        global_model.load_state_dict(copy.deepcopy(base_state))

        subset = [clients[i] for i in client_indices]
        for _ in range(budget_rounds):
            client_models = []
            client_sizes = []
            for (Xc, yc) in subset:
                ds = TabDataset(Xc, yc)
                dl = DataLoader(ds, batch_size=batch_size, shuffle=True)
                cm = MLP(input_dim, num_classes, hidden=hidden, dropout=dropout).to(device)
                cm.load_state_dict(copy.deepcopy(global_model.state_dict()))
                train_one_client(cm, dl, epochs=local_epochs, lr=lr, weight_decay=weight_decay)
                client_models.append(cm)
                client_sizes.append(len(yc))
            fedavg_aggregate(global_model, client_models, client_sizes)

        _, met = eval_model(global_model, X_val, y_val)
        return met["Accuracy"]

    shapley = np.zeros(n_clients, dtype=float)

    for _ in range(mc_permutations):
        perm = list(range(n_clients))
        random.shuffle(perm)

        prev_acc = 0.0
        subset = []
        for cid in perm:
            subset.append(cid)
            acc = train_subset(subset)
            shapley[cid] += (acc - prev_acc)
            prev_acc = acc

    shapley /= mc_permutations
    return shapley

# ---------------------------
# Core runner for one target
# ---------------------------
def run_federated_target(
    df, target_col,
    class_names=None,
    leakage_cols_extra=None,
    n_clients=5,
    rounds=20,
    local_epochs=1,
    lr=1e-3,
    weight_decay=1e-4,
    batch_size=128,
    hidden=128,
    dropout=0.25,
    do_fedshapley=True,
    fedshapley_perms=8,
    fedshapley_budget_rounds=3
):
    print("\n" + "="*60)
    print("TARGET:", target_col)
    print("="*60)

    data = df.copy()

    # Handle Category reconstruction if one-hot exists but Category doesn't
    if target_col == "Category" and "Category" not in data.columns:
        def reconstruct_category(row):
            if row.get('Category_Primary Cause', 0) == 1:
                return 'Primary Cause'
            elif row.get('Category_Trans-acting Modifier', 0) == 1:
                return 'Trans-acting Modifier'
            else:
                return 'Cis-acting Modifier'
        data["Category"] = data.apply(reconstruct_category, axis=1)

    # Encode target
    le_t = LabelEncoder()
    y = le_t.fit_transform(data[target_col].astype(str))

    if class_names is None:
        class_names = [str(c) for c in le_t.classes_]
    else:
        class_names = [str(x) for x in class_names]

    # Build features: drop leakage columns
    leakage_cols = ["Patient_ID", target_col]
    if leakage_cols_extra:
        leakage_cols += leakage_cols_extra

    X = data.drop(columns=leakage_cols, errors="ignore")

    # Encode any remaining object columns (if your preprocessed file already encoded, this does nothing)
    cat_cols = X.select_dtypes(include=["object", "category"]).columns
    for col in cat_cols:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))

    X = X.values.astype(np.float32)

    # 20% Train, 10% Val, 70% Test
    X_trainval, X_test, y_trainval, y_test = train_test_split(
        X, y, test_size=0.7, random_state=SEED, stratify=y
    )
    X_train, X_val, y_train, y_val = train_test_split(
        X_trainval, y_trainval, test_size=0.3333, random_state=SEED, stratify=y_trainval
    )

    input_dim = X_train.shape[1]
    num_classes = len(np.unique(y))

    # Split training set into clients
    clients = split_into_clients(X_train, y_train, n_clients=n_clients)

    # Init global model
    global_model = MLP(input_dim, num_classes, hidden=hidden, dropout=dropout).to(device)
    best_state = copy.deepcopy(global_model.state_dict())
    best_val_acc = -1.0

    train_acc_hist, val_acc_hist = [], []

    # Federated training (FedAvg)
    for r in range(1, rounds+1):
        client_models = []
        client_sizes = []

        for (Xc, yc) in clients:
            ds = TabDataset(Xc, yc)
            dl = DataLoader(ds, batch_size=batch_size, shuffle=True)

            cm = MLP(input_dim, num_classes, hidden=hidden, dropout=dropout).to(device)
            cm.load_state_dict(copy.deepcopy(global_model.state_dict()))

            train_one_client(cm, dl, epochs=local_epochs, lr=lr, weight_decay=weight_decay)

            client_models.append(cm)
            client_sizes.append(len(yc))

        fedavg_aggregate(global_model, client_models, client_sizes)

        # Evaluate each round: train + val
        _, train_met = eval_model(global_model, X_train, y_train)
        _, val_met   = eval_model(global_model, X_val, y_val)

        train_acc_hist.append(train_met["Accuracy"])
        val_acc_hist.append(val_met["Accuracy"])

        if val_met["Accuracy"] > best_val_acc:
            best_val_acc = val_met["Accuracy"]
            best_state = copy.deepcopy(global_model.state_dict())

        if r % max(1, rounds//5) == 0 or r == 1 or r == rounds:
            print(f"Round {r:>3}/{rounds} | Train Acc: {train_met['Accuracy']:.4f} | Val Acc: {val_met['Accuracy']:.4f}")

    # Load best model (early stopping by best val)
    global_model.load_state_dict(best_state)

    # Plot overfit/underfit curve
    plt.figure(figsize=(10,4))
    plt.plot(train_acc_hist, label="Train Accuracy")
    plt.plot(val_acc_hist, label="Val Accuracy")
    plt.xlabel("Round")
    plt.ylabel("Accuracy")
    plt.title(f"FedAvg Learning Curve ({target_col})")
    plt.legend()
    plt.show()

    print("\nDiagnosis:")
    print(overfit_underfit_diagnosis(train_acc_hist, val_acc_hist))

    # Final evaluations
    ypred_train, met_train = eval_model(global_model, X_train, y_train)
    ypred_val,   met_val   = eval_model(global_model, X_val, y_val)
    ypred_test,  met_test  = eval_model(global_model, X_test, y_test)
    ypred_full,  met_full  = eval_model(global_model, X, y)

    print("\nFINAL METRICS (Best Val Model)")
    print("Train:", met_train)
    print("Val  :", met_val)
    print("Test :", met_test)
    print("Full :", met_full)

    print("\nClassification Report (TEST):\n")
    print(classification_report(y_test, ypred_test, target_names=class_names, zero_division=0))

    # Confusion matrix (TEST) without seaborn to keep it simple
    cm = confusion_matrix(y_test, ypred_test)
    plt.figure(figsize=(7,5))
    plt.imshow(cm)
    plt.title(f"Confusion Matrix – FedAvg ({target_col})")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.xticks(range(len(class_names)), class_names, rotation=45, ha="right")
    plt.yticks(range(len(class_names)), class_names)
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, str(cm[i, j]), ha="center", va="center")
    plt.tight_layout()
    plt.show()

    # FedShapley contributions (approx)
    shapley = None
    if do_fedshapley:
        print("\nComputing FedShapley (approx) client contributions...")
        shapley = fedshapley_approx(
            X_train, y_train, X_val, y_val,
            input_dim=input_dim, num_classes=num_classes,
            n_clients=n_clients,
            mc_permutations=fedshapley_perms,
            budget_rounds=fedshapley_budget_rounds,
            local_epochs=1,
            lr=lr,
            weight_decay=weight_decay,
            batch_size=batch_size,
            hidden=hidden,
            dropout=dropout
        )
        # Print and plot
        for i, v in enumerate(shapley):
            print(f"Client {i}: Shapley contribution (ΔValAcc) ≈ {v:.4f}")

        plt.figure(figsize=(8,4))
        plt.bar([f"C{i}" for i in range(len(shapley))], shapley)
        plt.title(f"FedShapley (approx) – Client Contributions ({target_col})")
        plt.ylabel("Δ Validation Accuracy")
        plt.show()

    # Return results for comparison table
    rows = []
    def add_row(split_name, met):
        rows.append({
            "Target": target_col,
            "Split": split_name,
            "Accuracy": met["Accuracy"],
            "Precision": met["Precision"],
            "Recall": met["Recall"],
            "F1": met["F1"]
        })

    add_row("Train", met_train)
    add_row("Validation", met_val)
    add_row("Test", met_test)
    add_row("Full Dataset", met_full)

    return rows, shapley

# ---------------------------
# Run for all 3 targets
# ---------------------------
all_rows = []
all_shapleys = {}

# 1) Motor_Symptoms
rows, shap = run_federated_target(
    df=df,
    target_col="Motor_Symptoms",
    class_names=None,              # auto from label encoder
    leakage_cols_extra=None,
    n_clients=5,
    rounds=20,
    local_epochs=1,
    lr=1e-3,
    weight_decay=1e-4,
    batch_size=128,
    hidden=128,
    dropout=0.25,
    do_fedshapley=True,
    fedshapley_perms=8,
    fedshapley_budget_rounds=3
)
all_rows += rows
all_shapleys["Motor_Symptoms"] = shap

# 2) Disease_Stage
# If your dataset has exactly those 4 names, you can set them. Otherwise keep None.
rows, shap = run_federated_target(
    df=df,
    target_col="Disease_Stage",
    class_names=None,              # or ['Pre-Symptomatic','Early','Middle','Late'] if it matches your labels
    leakage_cols_extra=None
)
all_rows += rows
all_shapleys["Disease_Stage"] = shap

# 3) Category
rows, shap = run_federated_target(
    df=df,
    target_col="Category",
    class_names=None,              # or ['Primary Cause','Trans-acting Modifier','Cis-acting Modifier']
    leakage_cols_extra=None,
    n_clients=5,
    rounds=20,
    local_epochs=1,
    lr=1e-3,
    weight_decay=1e-4,
    batch_size=128,
    hidden=128,
    dropout=0.25,
    do_fedshapley=True,
    fedshapley_perms=8,
    fedshapley_budget_rounds=3
)
all_rows += rows
all_shapleys["Category"] = shap

# ---------------------------
# Comparison table
# ---------------------------
comparison_df = pd.DataFrame(all_rows)
print("\n" + "="*60)
print("FEDERATED LEARNING (FedAvg) PERFORMANCE COMPARISON")
print("="*60)
print(comparison_df.to_markdown(index=False))

# Display Shapley values if computed
print("\n" + "="*60)
print("FEDSHAPLEY (APPROX) CLIENT CONTRIBUTIONS")
print("="*60)
for target, shapley_values in all_shapleys.items():
    if shapley_values is not None:
        print(f"Target: {target}")
        for i, val in enumerate(shapley_values):
            print(f"  Client {i}: {val:.4f}")


```

```python
import numpy as np
import pandas as pd
import random
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ===== Load preprocessed dataset =====
path = "/content/drive/MyDrive/Dataset/Huntington_Disease_Dataset_Preprocessed_Scaled.csv.gz"
df = pd.read_csv(path)
print("Dataset loaded:", df.shape)

# -----------------------------
# Simple helpers
# -----------------------------
def build_model(input_dim, num_classes, lr=0.001, hidden1=128, hidden2=64, dropout=0.2):
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(input_dim,)),
        tf.keras.layers.Dense(hidden1, activation="relu"),
        tf.keras.layers.Dropout(dropout),
        tf.keras.layers.Dense(hidden2, activation="relu"),
        tf.keras.layers.Dense(num_classes, activation="softmax"),
    ])
    model.compile(
        optimizer=tf.keras.optimizers.Adam(lr),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

def get_metrics(y_true, y_pred):
    return {
        "Accuracy":  accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, average="weighted", zero_division=0),
        "Recall":    recall_score(y_true, y_pred, average="weighted", zero_division=0),
        "F1":        f1_score(y_true, y_pred, average="weighted", zero_division=0),
    }

def diagnose_fit(train_hist, val_hist):
    train_last, val_last = train_hist[-1], val_hist[-1]
    gap = train_last - val_last

    if train_last > 0.90 and gap > 0.08:
        return ("Likely OVERFITTING: train much higher than validation.\n"
                "Fixes: more dropout/L2, fewer hidden units, fewer rounds/local epochs, early stopping.")
    if train_last < 0.65 and val_last < 0.65 and abs(gap) < 0.03:
        return ("Likely UNDERFITTING: both train and validation are low and similar.\n"
                "Fixes: increase model capacity, train longer (more rounds/local epochs), tune LR, reduce regularization.")
    return ("Fit seems reasonable (train/val close).\n"
            "Improve via tuning LR/rounds/epochs, class balancing, better features.")

def print_classwise_check(y_true, y_pred, class_names):
    # Report as dict for easy parsing
    rep = classification_report(y_true, y_pred, target_names=[str(c) for c in class_names],
                                zero_division=0, output_dict=True)

    # Which class is predicted most?
    unique, counts = np.unique(y_pred, return_counts=True)
    pred_dist = dict(zip(unique, counts))

    # Find collapse: classes with recall==0
    never_predicted = []
    for i, name in enumerate(class_names):
        key = str(name)
        if key in rep and rep[key]["recall"] == 0.0:
            never_predicted.append(name)

    dominant_class = unique[np.argmax(counts)] if len(unique) else None

    print("\n--- CLASS-WISE CHECK ---")
    print("Predicted class distribution (by label index):", pred_dist)

    # Print table style
    print("\nPer-class metrics:")
    print(f"{'Class':>22} | {'Precision':>9} | {'Recall':>6} | {'F1':>6} | {'Support':>7}")
    print("-"*62)
    for name in class_names:
        key = str(name)
        p = rep[key]["precision"]
        r = rep[key]["recall"]
        f = rep[key]["f1-score"]
        s = int(rep[key]["support"])
        print(f"{str(name):>22} | {p:9.2f} | {r:6.2f} | {f:6.2f} | {s:7d}")

    # Interpretation like your sample
    if dominant_class is not None and len(never_predicted) >= (len(class_names) - 1):
        print("\n➡️ Model COLLAPSE detected:")
        print(f"- Dominant predicted class index: {dominant_class} (model predicts mostly ONE class)")
        print(f"- Never predicted / zero-recall classes: {never_predicted}")
    else:
        print("\n✅ No full collapse detected (model predicts multiple classes).")

def plot_confusion(cm, class_names, title):
    plt.figure(figsize=(7,5))
    plt.imshow(cm)
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.xticks(range(len(class_names)), [str(c) for c in class_names], rotation=45, ha="right")
    plt.yticks(range(len(class_names)), [str(c) for c in class_names])
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, str(cm[i, j]), ha="center", va="center")
    plt.tight_layout()
    plt.show()

def fedshapley_simple(clients, X_val, y_val, global_weights, build_fn,
                      perms=6, subset_epochs=1):
    """
    Approx FedShapley:
    - Utility = Val Accuracy after training sequentially on subset clients (quick approximation)
    """
    n = len(clients)
    shap = np.zeros(n, dtype=float)

    for _ in range(perms):
        order = list(range(n))
        random.shuffle(order)

        prev_u = 0.0
        subset = []

        for cid in order:
            subset.append(cid)

            temp_model = build_fn()
            temp_model.set_weights(global_weights)

            # train on subset clients quickly
            for j in subset:
                Xc, yc = clients[j]
                temp_model.fit(Xc, yc, epochs=subset_epochs, verbose=0)

            pred = np.argmax(temp_model.predict(X_val, verbose=0), axis=1)
            u = accuracy_score(y_val, pred)

            shap[cid] += (u - prev_u)
            prev_u = u

    return shap / perms

# -----------------------------
# Federated run with printing exactly like your output
# -----------------------------
def run_target(target_col, extra_drop=None,
               n_clients=5, rounds=20, local_epochs=1,
               lr=0.001, hidden1=128, hidden2=64, dropout=0.2,
               shap_perms=6, shap_subset_epochs=1):

    print("\n" + "="*60)
    print(f"TARGET: {target_col}")
    print("="*60)

    data = df.copy()

    # Handle Category reconstruction if needed
    if target_col == "Category" and "Category" not in data.columns:
        def reconstruct_category(row):
            if row.get("Category_Primary Cause", 0) == 1:
                return "Primary Cause"
            elif row.get("Category_Trans-acting Modifier", 0) == 1:
                return "Trans-acting Modifier"
            else:
                return "Cis-acting Modifier"
        data["Category"] = data.apply(reconstruct_category, axis=1)

    # Encode target
    le = LabelEncoder()
    y = le.fit_transform(data[target_col].astype(str))
    class_names = le.classes_

    # Features
    drop_cols = ["Patient_ID", target_col]
    if extra_drop:
        drop_cols += extra_drop
    X = data.drop(columns=drop_cols, errors="ignore")

    # If any object columns exist
    for c in X.select_dtypes(include=["object"]).columns:
        X[c] = LabelEncoder().fit_transform(X[c].astype(str))

    X = X.values.astype(np.float32)

    # Split (20 train, 10 val, 70 test)
    X_trainval, X_test, y_trainval, y_test = train_test_split(
        X, y, test_size=0.7, stratify=y, random_state=SEED
    )
    X_train, X_val, y_train, y_val = train_test_split(
        X_trainval, y_trainval, test_size=0.3333, stratify=y_trainval, random_state=SEED
    )

    # Clients from TRAIN
    idx = np.arange(len(y_train))
    np.random.shuffle(idx)
    splits = np.array_split(idx, n_clients)
    clients = [(X_train[s], y_train[s]) for s in splits]

    input_dim = X_train.shape[1]
    num_classes = len(np.unique(y))

    def build_fn():
        return build_model(input_dim, num_classes, lr=lr, hidden1=hidden1, hidden2=hidden2, dropout=dropout)

    # FedAvg training
    global_model = build_fn()
    best_weights = global_model.get_weights()
    best_val_acc = -1

    train_hist, val_hist = [], []

    for r in range(1, rounds+1):
        client_weights = []
        client_sizes = []

        for Xc, yc in clients:
            local_model = build_fn()
            local_model.set_weights(global_model.get_weights())
            local_model.fit(Xc, yc, epochs=local_epochs, verbose=0)
            client_weights.append(local_model.get_weights())
            client_sizes.append(len(yc))

        # FedAvg aggregation
        new_weights = []
        for weights in zip(*client_weights):
            new_weights.append(np.average(weights, axis=0, weights=client_sizes))
        global_model.set_weights(new_weights)

        # Evaluate train/val acc
        tr_pred = np.argmax(global_model.predict(X_train, verbose=0), axis=1)
        va_pred = np.argmax(global_model.predict(X_val, verbose=0), axis=1)

        tr_acc = accuracy_score(y_train, tr_pred)
        va_acc = accuracy_score(y_val, va_pred)

        train_hist.append(tr_acc)
        val_hist.append(va_acc)

        if va_acc > best_val_acc:
            best_val_acc = va_acc
            best_weights = global_model.get_weights()

        if r in [1, 4, 8, 12, 16, rounds] or r % 4 == 0:
            print(f"Round {r:>3}/{rounds} | Train Acc: {tr_acc:.4f} | Val Acc: {va_acc:.4f}")

    # Load best val model
    global_model.set_weights(best_weights)

    # Diagnosis
    print("\n\nDiagnosis:")
    print(diagnose_fit(train_hist, val_hist))

    # Final predictions
    pred_train = np.argmax(global_model.predict(X_train, verbose=0), axis=1)
    pred_val   = np.argmax(global_model.predict(X_val, verbose=0), axis=1)
    pred_test  = np.argmax(global_model.predict(X_test, verbose=0), axis=1)
    pred_full  = np.argmax(global_model.predict(X, verbose=0), axis=1)

    m_tr = get_metrics(y_train, pred_train)
    m_va = get_metrics(y_val, pred_val)
    m_te = get_metrics(y_test, pred_test)
    m_full = get_metrics(y, pred_full)

    print("\nFINAL METRICS (Best Val Model)")
    print("Train:", m_tr)
    print("Val  :", m_va)
    print("Test :", m_te)
    print("Full :", m_full)

    print("\nClassification Report (TEST):\n")
    print(classification_report(y_test, pred_test, target_names=[str(c) for c in class_names], zero_division=0))

    # --- CLASS-WISE CHECK LIKE YOU WANT ---
    print_classwise_check(y_test, pred_test, class_names)

    # Confusion matrix plot
    cm = confusion_matrix(y_test, pred_test)
    plot_confusion(cm, class_names, f"Confusion Matrix — {target_col}")

    # Curves plot
    plt.figure(figsize=(9,4))
    plt.plot(train_hist, label="Train Acc")
    plt.plot(val_hist, label="Val Acc")
    plt.title(f"Train vs Val Accuracy — {target_col}")
    plt.xlabel("Round")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()

    # FedShapley
    print("\nComputing FedShapley (approx) client contributions...")
    shap = fedshapley_simple(
        clients=clients,
        X_val=X_val,
        y_val=y_val,
        global_weights=global_model.get_weights(),
        build_fn=build_fn,
        perms=shap_perms,
        subset_epochs=shap_subset_epochs
    )

    for i, v in enumerate(shap):
        print(f"Client {i}: Shapley contribution (ΔValAcc) ≈ {v:.4f}")

    best_client = int(np.argmax(shap))
    print(f"\nBest contributing client: Client {best_client} (highest Shapley)")

    return {
        "target": target_col,
        "metrics": {"train": m_tr, "val": m_va, "test": m_te, "full": m_full},
        "shapley": shap
    }

# -----------------------------
# RUN ALL 3 TARGETS
# -----------------------------
results = []

results.append(run_target("Motor_Symptoms"))

results.append(run_target("Disease_Stage"))

# Category: drop one-hot leakage columns if they exist
results.append(run_target("Category", extra_drop=["Category_Primary Cause", "Category_Trans-acting Modifier"]))

print("\n" + "="*60)
print("DONE — You now have class-wise checks + shapley for each target")
print("="*60)
```
